```python
# Step 29 : Numerical Feature Distribution Summary

numerical_summary = train_df[numerical_columns].describe().T

numerical_summary
```

```python
# Step 30 : Categorical Feature Distribution

categorical_features = [
    "Gender",
    "Customer Type",
    "Type of Travel",
    "Class"
]

for column in categorical_features:
    print(f"\n{'=' * 60}")
    print(f"{column}")
    print(f"{'=' * 60}")
    print(train_df[column].value_counts())
```

```python
# Step 31 : Target Variable Distribution

target_distribution = train_df[target_column].value_counts()

print("Target Variable Distribution:")
print(target_distribution)

print("\nTarget Variable Percentage:")
print(
    (train_df[target_column].value_counts(normalize=True) * 100).round(2)
)
```

```python
# Step 32 : Test Target Variable Distribution

test_target_distribution = test_df[target_column].value_counts()

print("Test Target Variable Distribution:")
print(test_target_distribution)

print("\nTest Target Variable Percentage:")
print(
    (test_df[target_column].value_counts(normalize=True) * 100).round(2)
)
```

```python
# Step 33 : Numerical Features and Target Relationship

numerical_features_for_analysis = [
    column for column in numerical_columns
    if column not in ["Unnamed: 0", "id"]
]

for column in numerical_features_for_analysis:
    print(f"\n{'=' * 60}")
    print(f"{column}")
    print(f"{'=' * 60}")
    
    print(
        train_df.groupby(target_column)[column]
        .mean()
        .round(2)
    )
```

```python
# Step 34 : Categorical Features and Target Relationship

for column in categorical_features:
    print(f"\n{'=' * 60}")
    print(f"{column} vs {target_column}")
    print(f"{'=' * 60}")
    
    print(
        pd.crosstab(
            train_df[column],
            train_df[target_column],
            normalize="index"
        ).round(3)
    )
```

```python
# Step 35 : Feature Data Types

data_types_summary = pd.DataFrame({
    "Column": train_df.columns,
    "Data Type": train_df.dtypes.astype(str)
})

data_types_summary
```

```python
# Step 36 : Unique Value Count for All Features

unique_value_summary = pd.DataFrame({
    "Column": train_df.columns,
    "Unique Values": train_df.nunique()
}).sort_values(
    by="Unique Values",
    ascending=True
)

unique_value_summary
```

```python
# Step 37 : Potential Outlier Analysis

outlier_summary = []

for column in numerical_features_for_analysis:
    Q1 = train_df[column].quantile(0.25)
    Q3 = train_df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = (
        (train_df[column] < lower_bound) |
        (train_df[column] > upper_bound)
    ).sum()
    
    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": round(
            outlier_count / len(train_df) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by="Outlier Count",
    ascending=False
)
```

```python
# Step 38 : Train-Test Feature Consistency

train_columns = set(train_df.columns)
test_columns = set(test_df.columns)

print("Columns only in Training Dataset:")
print(train_columns - test_columns)

print("\nColumns only in Testing Dataset:")
print(test_columns - train_columns)

print("\nSame Columns:", train_columns == test_columns)
```

```python
# Step 39 : Train-Test Numerical Range Comparison

range_comparison = pd.DataFrame({
    "Train Min": train_df[numerical_features_for_analysis].min(),
    "Train Max": train_df[numerical_features_for_analysis].max(),
    "Test Min": test_df[numerical_features_for_analysis].min(),
    "Test Max": test_df[numerical_features_for_analysis].max()
})

range_comparison
```

```python
# Step 40 : Identify Non-Predictive Identifier Columns

identifier_columns = [
    "Unnamed: 0",
    "id"
]

print("Columns identified for removal during preprocessing:")
print(identifier_columns)
```
